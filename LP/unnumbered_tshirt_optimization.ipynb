# T-shirt Production Optimization

```python
from pulp import LpProblem, LpMaximize, LpVariable, LpInteger, PULP_CBC_CMD, LpStatus
```

```python
problem1 = LpProblem(name="problem", sense=LpMaximize)
T = LpVariable(name="nb_of_tshirt", lowBound=0, cat=LpInteger)
S = LpVariable(name="nb_of_sshirt", lowBound=0, cat=LpInteger)

problem1 += 25 * T + 20 * S, "objective_function"
problem1 += 30 * T + 15 * S <= 3360, "time_constraint"
problem1 += 0.2 * T + 0.5 * S <= 40, "volume_constraint"
```

```python
problem1.solve(solver=PULP_CBC_CMD(msg=0))

print(f"Status: {LpStatus[problem1.status]}")
if LpStatus[problem1.status] != "Optimal":
    raise RuntimeError("An optimal solution was not found.")

print(f"T: {T.value():.0f}")
print(f"S: {S.value():.0f}")
print(f"Objective: {problem1.objective.value():.0f}")
```
