# 9. Cee's Candy Company

```python
import pulp as pl

def solve(model):
    model.solve(pl.PULP_CBC_CMD(msg=False))
    if pl.LpStatus[model.status] != "Optimal":
        raise RuntimeError(pl.LpStatus[model.status])

def rainbow(exact=True):
    model = pl.LpProblem("Rainbow_Box", pl.LpMinimize)
    C = pl.LpVariable("chocolates", lowBound=0)
    P = pl.LpVariable("pastels", lowBound=0)
    model += 4*C + P
    model += .15*C + .05*P >= .10
    model += .5*C + P >= .60
    model += .5*C >= .20
    if exact:
        model += C + P == 1
    else:
        model += C + P >= 1
    solve(model)
    print(f"{'Exactly' if exact else 'At least'} one pound: chocolates = {C.value():.6f} lb; pastels = {P.value():.6f} lb; minimum cost = USD {pl.value(model.objective):.2f}")

rainbow(exact=True)
rainbow(exact=False)
```

```python
model = pl.LpProblem("Candy_Production", pl.LpMaximize)
F = pl.LpVariable("creamy_fudge", lowBound=0)
B = pl.LpVariable("burgundy_cherry", lowBound=0)
model += 2*F + B
model += .5*F + .3*B <= 500, "pecans"
model += 2*F + 2*B <= 3000, "chocolate"
model += 3*F + B <= 2400, "vanilla"
model += 24*B <= 28000, "cherries"
model += B >= .5*F, "minimum_cherry_ratio"
solve(model)
print(f"Creamy fudge: {F.value():.6f} lb")
print(f"Burgundy cherry: {B.value():.6f} lb")
print(f"Maximum contribution: USD {pl.value(model.objective):,.6f}")
```

Exactly 1 lb: **0.5 lb chocolates + 0.5 lb pastels**, costing **$2.50**.

At least 1 lb: **0.4 lb chocolates + 0.8 lb pastels**, costing **$2.40**.

Produce **550 lb creamy fudge** and **750 lb Burgundy cherry**. Maximum contribution: **$1,850**.
