# 5. Financial Portfolio

```python
import pulp as pl

def solve(model):
    model.solve(pl.PULP_CBC_CMD(msg=False))
    if pl.LpStatus[model.status] != "Optimal":
        raise RuntimeError(pl.LpStatus[model.status])

names = ["First mortgages", "Second mortgages", "Personal loans", "Commercial loans", "Government securities"]
returns = [0.09, 0.12, 0.15, 0.08, 0.06]
risks = [3, 6, 8, 2, 1]
model = pl.LpProblem("Financial_Portfolio", pl.LpMaximize)
x = [pl.LpVariable(f"investment_{i}", lowBound=0) for i in range(5)]
U = pl.LpVariable("savings", lowBound=0)
I = pl.lpSum(x)
model += pl.lpSum(returns[i]*x[i] for i in range(5)) + 0.03*U
model += I + U == 100
model += pl.lpSum(risks[i]*x[i] for i in range(5)) <= 5*I
model += x[3] >= 0.2*I
model += x[1] + x[2] <= x[0]
solve(model)
for name, v in zip(names, x):
    print(f"{name}: USD {v.value()*1e6:,.2f}")
print(f"Savings: USD {U.value()*1e6:,.2f}")
print(f"Annual return: USD {pl.value(model.objective)*1e6:,.2f}")
print(f"Return per budget dollar: {pl.value(model.objective)/100:.2%}")
print(f"Average investment risk: {sum(risks[i]*x[i].value() for i in range(5))/pl.value(I):.6f}")
```

Invest **$40 million in first mortgages**, **$40 million in personal loans**, and **$20 million in commercial loans**.

Annual return: **$11.2 million (11.2%)**. Average investment risk: **4.8**.
