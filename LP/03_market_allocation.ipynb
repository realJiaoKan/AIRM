# 3. A Market Allocation Problem

```python
import pulp as pl

def solve(model):
    model.solve(pl.PULP_CBC_CMD(msg=False))
    if pl.LpStatus[model.status] != "Optimal":
        raise RuntimeError(pl.LpStatus[model.status])

def advertising(men_target=24):
    model = pl.LpProblem("Market_Allocation", pl.LpMinimize)
    T = pl.LpVariable("TV_minutes", lowBound=0)
    M = pl.LpVariable("magazine_pages", lowBound=0)
    model += 600*T + 500*M
    model += 5*T + 2*M >= 24, "boys"
    model += T + 6*M >= 18, "women"
    model += 3*T + 3*M >= men_target, "men"
    solve(model)
    return model, T, M

model, T, M = advertising()
print(f"TV minutes: {T.value():.6f}")
print(f"Magazine pages: {M.value():.6f}")
print(f"Minimum cost: {pl.value(model.objective):,.6f}")
extra, _, _ = advertising(men_target=25)
print(f"Cost of reaching an extra million retired men: {pl.value(extra.objective)-pl.value(model.objective):.6f}")
radio_value = sum(c.pi for c in model.constraints.values())
print(f"Maximum marginal price of a radio spot: {radio_value:.6f}")
```

Purchase **2.6667 TV minutes** and **5.3333 magazine pages**. Minimum cost: **4,266.6667** in the table’s cost units.

Reaching an extra million retired men costs **144.4444**. The maximum marginal price of one radio spot is **177.7778**.
