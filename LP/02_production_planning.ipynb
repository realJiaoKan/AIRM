# Production Planning

```python
from pulp import LpProblem, LpMaximize, LpVariable, PULP_CBC_CMD, LpStatus, value

model = LpProblem("Weekly_Production_Planning", LpMaximize)
x = LpVariable("product_1", lowBound=0, upBound=130)
y = LpVariable("product_2", lowBound=0, upBound=100)

model += 30 * x + 45 * y, "weekly_profit"
model += 1.5 * x + 2 * y <= 300, "production_hours"
model += 0.25 * (x + y) <= 45, "hand_finishing_hours"

model.solve(PULP_CBC_CMD(msg=0))
if LpStatus[model.status] != "Optimal":
    raise RuntimeError("An optimal solution was not found.")

print(f"Product 1: {x.value():.4f} units per week")
print(f"Product 2: {y.value():.4f} units per week")
print(f"Maximum weekly objective: GBP {value(model.objective):,.2f}")
print(f"Component A: {x.value():.4f} units")
print(f"Component B: {2 * y.value():.4f} units")
print(f"Component C: {x.value() + y.value():.4f} units")
```

Produce **66.6667 units of product 1** and **100 units of product 2** per week. Maximum weekly objective: **£6,500**.
