# 4. Workforce Planning

```python
import pulp as pl

def solve(model):
    model.solve(pl.PULP_CBC_CMD(msg=False))
    if pl.LpStatus[model.status] != "Optimal":
        raise RuntimeError(pl.LpStatus[model.status])

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
required = [14, 13, 15, 16, 19, 18, 11]
model = pl.LpProblem("Workforce_Planning", pl.LpMinimize)
x = [pl.LpVariable(f"start_{day}", lowBound=0, cat=pl.LpInteger) for day in days]
model += pl.lpSum(x)
for d in range(7):
    model += pl.lpSum(x[(d-k) % 7] for k in range(5)) >= required[d], f"coverage_{days[d]}"
solve(model)
print(f"Minimum workforce: {pl.value(model.objective):.0f}")
print("Day          Start   Working   Required")
for d, day in enumerate(days):
    working = sum(x[(d-k) % 7].value() for k in range(5))
    print(f"{day:12} {x[d].value():5.0f} {working:9.0f} {required[d]:10}")
```

Minimum workforce: **22 workers**.

Workers starting Monday through Sunday: **4, 6, 2, 7, 0, 3, 0**.
